### Option 2: Fill the missing values with scikit learn

```python
car_missing_data = pd.read_csv("data/car-sales-extended-missing-data.csv")
car_missing_data.head()
```

```python
car_missing_data.isna().sum()
```

```python
car_missing_data.info()
```

```python
# Now we want to remove the missing values from the label
car_missing_data.dropna(subset="Price", inplace=True)
```

```python
car_missing_data.isna().sum()
```

```python
# Now let's try to convert these data in features and labels
X = car_missing_data.drop("Price", axis=1)

Y = car_missing_data["Price"]
```

```python
X.head()
```

```python
# Now will see how to fill missing values with imputation and transformer
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer

# Now we have to make imputation feature
cat_imputer = SimpleImputer(strategy="constant", fill_value="missing")
door_imputer = SimpleImputer(strategy="constant", fill_value=4)
num_imputer = SimpleImputer(strategy="mean")

# Now we have to seperate the column's
cat_features = ["Make", "Colour"]
door_features = ["Doors"]
num_features = ["Odometer (KM)"]

# Now let's build the transformer
transformer = ColumnTransformer(
    transformers=[
        ("cat_imputer", cat_imputer, cat_features),
        ("door_imputer", door_imputer, door_features),
        ("num_imputer", num_imputer, num_features)
])

# Now we have to transform the data
x_transformed = transformer.fit_transform(X)
x_transformed
```

```python
car_sales_filled = pd.DataFrame(x_transformed, columns=["Make", "Colour", "Doors", "Odometer (KM)"])
car_sales_filled.head()
```

```python
car_sales_filled.isna().sum()
```

```python
# Now we have to convert these values into numbers
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Now we have to define the features we want to convert into numbers
feature_columns = ["Make", "Colour"]

# We have to define the encoder also
one_hot = OneHotEncoder()

# Now we have to define the transformer
transformer = ColumnTransformer(
    transformers=[
        ("one_hot", one_hot, feature_columns)
    ],
    remainder="passthrough"
)

# Now we have to transform our data
x_transformed = transformer.fit_transform(car_sales_filled)
x_transformed
```

```python
# Now let's fit a model...
np.random.seed(42)
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split

# Now we have to split the data into train and test
x_train, x_test, y_train, y_test = train_test_split(x_transformed, Y, test_size=0.2)

# Now we have to train our model
model = RandomForestRegressor()

model.fit(x_train, y_train)

model.score(x_test, y_test)
```

* Let's pick one of the `ensamble methods`

  * Basically it gives the average result of all of the algorithms so that we get more efficiet presdictions.

> So basically the RandomForest ensamble estimator gives predicction based on several base decision trees output.

- Decision trees are the building block of RandomForest.

    - For regression task RandomForest predicts by making an `average` of the base trees predictions

## 4.2 the score paremeter:
We can use cross validation evaluation matrix in here. Now what is this?
- Basically `cross validation` is used to define how our model performs in different train and test split. 
- There is a parameter called cv=`which define the splitting number`
- gives us the performens on different splits

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
np.random.seed(42)
# first we have to define the model
model = RandomForestClassifier()

# Now we have to split in features and labels
X = heart_disease.drop("target", axis=1)
Y = heart_disease["target"]

# Now we have to split our data in train and test 
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)

# Now we have to train our model with trainig dataset
model.fit(x_train, y_train);
```

```python
# Now as we have set the cross val score previously so we can split our data differntly
np.random.seed(42)
score_from_cvs = cross_val_score(model, X,Y, cv=5)
score_from_cvs
```

```python
# Now let's increase the cv number which denotes to splits
np.random.seed(42)
cross_val_score(model, X,Y, cv=10)
```

```python
# Now as we get cross valued scores. we can make an average from them and compare with our direct score that model gives
model_score = model.score(x_test, y_test)
model_score, np.mean(score_from_cvs).item()
```

### Area under the Receiver Operating Charectaristic Curve (ROC curve)

*ROC curve measures True positive rate Vs False possitive rate*

- True positive: if prediction is `1` truth is `1`
- False positive: if pediction is `1` truth is `0`
- True negative: if prediction is `0` truth is `0`
- False negative: if prediction is `0` truth is `1`


```python
heart_disease.head()
```

```python
from sklearn.metrics import roc_curve

np.random.seed(42)
# Split the data into train test
x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2)

# Now we have to fit the model with training data
model.fit(x_train, y_train)

# Now let's see the prediction of this
y_proba = model.predict_proba(x_test)
y_proba[:10]
```

```python
model.predict(x_test)[:10]
```

```python
# Now we will calculate the positive probability
y_positive_proba = y_proba[:,1]
y_positive_proba[:10]
```

```python
# Now we have to calculate fpr, tpr, thresholds
fpr, tpr, thresholds = roc_curve(y_test, y_positive_proba)
```

```python
fpr
```

```python
tpr
```

```python
#  Now we have to plot a roc curve
import matplotlib.pyplot as plt

def plot_roc_curve(fpr, tpr):
    plt.plot(fpr, tpr, color="orange", label="ROC")
    # plotting the base line
    plt.plot([0,1],[0,1], color="blue", linestyle="--", label="guessing")

    # Customizing the plot
    plt.xlabel("False positive rate")
    plt.ylabel("True positive rate")
    plt.legend()
    plt.show()

plot_roc_curve(fpr,tpr)
```

```python
# Now we have to see the AUC `Area under the curve`
from sklearn.metrics import roc_auc_score
roc_auc_score(y_test, y_positive_proba)
```

```python
# Now let's see how a perfect ROC curve looks like
fpr, tpr, thresholds = roc_curve(y_test, y_test)
fpr
```

```python
plot_roc_curve(fpr,tpr)
```

```python
# Now let's see the area under the curve. it will be the highest value:1
roc_auc_score(y_test, y_test)
```
