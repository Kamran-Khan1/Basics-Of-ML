## Varients and Standerd Diviation

Variants: a measure of how spread out the distribution of numbers are <br/>
High variant: means the numbers are spread far <br/>
Low variant: means the numbers are clustered closely

```python
# see this in practice
High_vlaue_array = np.array([1,100,200,1000,100000])
Low_value_array = np.array([2,4,6,8,10])
```

```python
np.var(High_vlaue_array), np.var(Low_value_array)
```

Standard Diviation: a measure of the amount of variation or dispersion of a set of values.

```python
np.std(High_vlaue_array), np.std(Low_value_array)
```

```python
np.mean(High_vlaue_array), np.mean(Low_value_array)
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.hist(High_vlaue_array)
plt.show()
```

```python
plt.hist(Low_value_array)
plt.show()
```

## Dot product and matrix multiplication

```python
np.random.seed(0)
mat1 = np.random.randint(10, size=(5,3))
mat2 = np.random.randint(10, size=(5,3))
```

```python
mat1
```

```python
mat2
```

```python
# Matrix multiplicatoin
mat1 * mat2
```

```python
mat1.shape, mat2.shape
```

```python
# So normal multiplication multiplies the elements individually
# But dot product is different (RxC) = (RxC) 1st Matrix column and second matrix Row should match
```

```python
# np.dot(mat1, mat2) # This will give a valueError, cause not convay with the rules
```

```python
# So the solution is to match the row and columns of the matrix
mat3 = mat2.T #Let's transpose the matrix, so it will be (3x5) 
# Now it is (5x3) * (3x5)
```

```python
dot_product = np.dot(mat1, mat3)
```

```python
dot_product
```

```python
dot_product.shape
```
