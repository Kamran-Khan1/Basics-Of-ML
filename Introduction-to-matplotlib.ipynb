## Matplotlib basic workflow

```python
# 0. We have to import matplotlib 
%matplotlib inline
import matplotlib.pyplot as plt

# 1. prepare data
x = [1,2,3,4]
y = [11,7,9,14]

# 2. setup plot
fig, ax = plt.subplots(figsize=(8,5))

# 3. plot the data
ax.plot(x,y)

# 4. customize plot
ax.set(title="simple plot",
      xlabel= "x-axis",
      ylabel= "y-axis")
# 5. save and show the plot
fig.savefig("images/simple-plot.png")
plt.show();
```

```python
# Bar graphs
gadgets_store = {"Mobile": 100, "Laptop":312, "Watch(d)": 80}
fig, ax = plt.subplots()
ax.bar(gadgets_store.keys(), gadgets_store.values())
ax.set(title="Kamran's Gadgets",
      ylabel="price ($)")
plt.savefig("images/gadgets.png")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.barh(gadgets_store.keys(), gadgets_store.values()) #This means bar in a horizontal position
plt.show()
```

```python
# Let's see some histograms
x = np.random.randn(10000)
fig, ax = plt.subplots() # Normal distribution of 10000 numbers
ax.hist(x)
plt.show()
```

### Two options for subplots

```python
# Option 1
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,
                                         ncols=2,
                                         figsize=(7,5))
ax1.plot(x, x/2)
ax2.scatter(np.random.random(10), np.random.random(10))
ax3.bar(gadgets_store.keys(), gadgets_store.values())
ax4.hist(x)
plt.show()
```

```python
# Option 2 

fig, ax = plt.subplots(nrows=2,
                      ncols=2,
                      figsize=(7,5))
ax[0,0].plot(x, x/2)
ax[0,1].scatter(np.random.random(10), np.random.random(10))
ax[1,0].bar(gadgets_store.keys(), gadgets_store.values())
ax[1,1].hist(x)
plt.savefig("images/subplots.png")
plt.show()
```

### Plotting differences between normal plot Vs OO plot 
* Use normal plot when you need quick plots
* Use OO plot when you need complex plots

```python
heart_disease.head()
```

```python
# let's try to extract the age greater then 50
above_50 = heart_disease[heart_disease["age"] > 50]
above_50.head()
```

```python
above_50.shape
```

```python
# Let's make a scatter plot with plot method
above_50.plot.scatter(x="age",
                     y= "chol",
                     c= "target");
```

```python
# Let's do this same with OO plots
fig, ax = plt.subplots(figsize=(10,6))
above_50.plot.scatter(x="age",
             y="chol",
             c="target",
             ax=ax);
```

```python
# Now let's go for the actual OO plots
#1. define the subplots
fig, ax = plt.subplots(figsize=(10,6))

#2. plot the data
scatter = ax.scatter(x=above_50["age"],
                    y=above_50["chol"],
                    c=above_50["target"])
#3. Let's customise our plot
ax.set(title="Heart disease scatter plot",
      xlabel="Age",
      ylabel="Cholestrol")
#4. Let's add some lagends
ax.legend(*scatter.legend_elements(),title="Target")

#5. add horizontal line of mean cholestrol
ax.axhline(np.mean(above_50["chol"]), linestyle="--");
```
